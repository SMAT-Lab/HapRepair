```python
import requests
from bs4 import BeautifulSoup
import json

def scrape_harmonyos_page(url):
    response = requests.get(url)
    response.encoding = 'utf-8'
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        print(soup)
        
        # Find the rule sections (adjust based on actual HTML structure)
        rules = []
        for rule_section in soup.find_all('div', class_='layout-right'):  # Adjust the class name as necessary
            print(rule_section)
            rule_name = rule_section.find('h1').get_text(strip=True)  # Assuming rule name is in h2 tag
            rule_description = rule_section.find('p').get_text(strip=True)  # Assuming description is in p tag
            code_example_tag = rule_section.find('pre')
            code_example = code_example_tag.get_text(strip=True) if code_example_tag else "No code example available"
            
            rules.append({
                "name": rule_name,
                "description": rule_description,
                "code_example": code_example.replace('\n', '\\n').replace('"', '\\"')
            })

        # Output the scraped rules in the desired format
        return json.dumps(rules, indent=4, ensure_ascii=False)
    else:
        return f"Failed to retrieve page, status code: {response.status_code}"

# URL to scrape
url = "https://developer.huawei.com/consumer/cn/doc/harmonyos-guides-V5/ide_hp-arkui-use-row-column-to-replace-flex-V5"

# Call the function and print the result
rules_json = scrape_harmonyos_page(url)
print(rules_json)
```
