```python
### 统计ets，ts文件总个数
import os

def count_ts_files(directory):
    ts_count = 0
    ets_count = 0
    
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.ts'):
                ts_count += 1
            elif file.endswith('.ets'):
                ets_count += 1
                
    print(f'找到 .ts 文件: {ts_count} 个')
    print(f'找到 .ets 文件: {ets_count} 个')
    print(f'总计: {ts_count + ets_count} 个文件')
    
    return ts_count, ets_count

# 使用示例:
directory = './'  # 当前目录
ts_files, ets_files = count_ts_files(directory)
```
