```python
import os

def get_ets_files(directory):
    ets_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.ts'):
                ets_files.append(os.path.join(root, file))
    return ets_files

# 获取deepseek_origin/round_0目录下的所有.ts文件
directory = 'deepseek_origin/round_0'
ets_files = get_ets_files(directory)

print(f"找到 {len(ets_files)} 个.ts文件:")
ts_files_dir = 'ts_files'
if not os.path.exists(ts_files_dir):
    os.makedirs(ts_files_dir)

for i, file in enumerate(ets_files):
    print(f"正在处理第 {i+1} 个文件: {file}")
    # 将文件内容读取到内存中
    with open(file, 'r', encoding='utf-8') as f:
        file_content = f.read()
    # 将文件内容写入到ts_files目录下的文件中
    with open(os.path.join(ts_files_dir, f"{i}.ts"), 'w', encoding='utf-8') as f:
        f.write(file_content)
```
