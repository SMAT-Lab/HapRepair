````python
import glob
import os
import pandas as pd

round_num = 5
proj_dir = f"./round_{round_num}"
result_files = glob.glob(os.path.join(proj_dir, '**', 'result*.xlsx'), recursive=True)
for result_file in result_files:
    df = pd.read_excel(result_file, skiprows=1)
    error_df = df[df['Severity'] == 'error']

    # 遍历每个error文件
    for _, row in error_df.iterrows():
        file_path = os.path.normpath(row['Source File'])
        # 从完整路径中提取相对路径部分
        relative_path = file_path.split(f'ets\\llama-3.2-90b-vision-instruct\\round_{round_num}\\')[-1] if '\\' in file_path else file_path.split(f'ets/llama-3.2-90b-vision-instruct/round_{round_num}/')[-1]
        relative_path = relative_path.replace('\\', '/')
        
        # 构建round_0和round_1中的文件路径
        round_0_path = f"./round_{round_num-1}/{relative_path}"
        round_1_path = f"./round_{round_num}_reject/{relative_path}"
        
        # 确保round_0中的源文件存在
        if os.path.exists(round_0_path):
            # 创建round_1文件所在的目录(如果不存在)
            os.makedirs(os.path.dirname(round_1_path), exist_ok=True)
            
            # 从round_0复制文件到round_1
            with open(round_0_path, 'r', encoding='utf-8') as src:
                content = src.read()
            with open(round_1_path, 'w', encoding='utf-8') as dst:
                dst.write(content)
            print(f"已将 {round_0_path} 复制到 {round_1_path}")
        else:
            print(f"源文件不存在: {round_0_path}")



````
